```python
import matplotlib.pyplot as plt
import seaborn as sns
```

Q1. What is the distribution of house prices?

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 5))
sns.histplot(data=df, x='price', bins=30, kde=True)

plt.title('Distribution of House Prices')
plt.xlabel('House Price')
plt.ylabel('Number of Houses')

plt.show()
```

Q2. What is the average price of houses?

```python
average_price = df['price'].mean()

print("Average House Price:", average_price)
```

Q3. How does the number of bedrooms affect house prices?

```python
plt.figure(figsize=(10, 5))

sns.barplot(data=df, x='bedrooms', y='price')

plt.title('Bedrooms vs House Prices')
plt.xlabel('Number of Bedrooms')
plt.ylabel('Average House Price')

plt.show()
```

Q4. What is the relationship between living area and house prices?

```python
plt.figure(figsize=(10, 5))

sns.scatterplot(data=df, x='sqft_living', y='price')

plt.title('Living Area vs House Prices')
plt.xlabel('Living Area (Square Feet)')
plt.ylabel('House Price')

plt.show()
```

Q5. How does the number of bathrooms affect house prices?

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('/content/kc_house_data_NaN.csv')

plt.figure(figsize=(10, 5))

sns.barplot(data=df, x='bathrooms', y='price', errorbar=None)

plt.title('Bathrooms vs House Prices')
plt.xlabel('Number of Bathrooms')
plt.ylabel('Average House Price')

plt.xticks(rotation=45)
plt.show()
```
